#bookings

```python

# Read the Excel file
file_path = "D:/Guvi/Project3/Data/bookings.csv"
bookings = pd.read_csv(file_path)

# Display first few rows
bookings.head()
```

```python
bookings.shape
```

```python
bookings.info()
```

```python
# Check for duplicates
bookings.duplicated().sum()
```

```python
bookings.columns
```

```python
#Step 2: Check for null values
print(bookings.isnull().sum())
```

```python
bookings.isnull().sum()
```

```python
# Fix dtypes - use errors='coerce' for dirty data
bookings['booking_date'] = pd.to_datetime(bookings['booking_date'], errors='coerce').dt.date
bookings['booking_time'] = pd.to_datetime(bookings['booking_time'], errors='coerce').dt.time
bookings['is_weekend'] = bookings['is_weekend'].astype(bool)

# This line replaces ALL NaN with None for MySQL. Don't use fillna(None)
bookings['incomplete_ride_reason'] = bookings['incomplete_ride_reason'].fillna("Others")
bookings['actual_ride_time_min'] = bookings['actual_ride_time_min'].fillna(0)
```

```python
bookings.dtypes
```

```python
print(bookings.describe(include='all').T)
print(bookings.describe().T)
```

```python
#Step 9: Creating a Table in the Database
cursor_mysql.execute("USE guvi_db;")  # Select database

# Drop if exists so we can recreate with correct syntax
cursor_mysql.execute("DROP TABLE IF EXISTS bookings;")
```

```python
bookings.info()
```

```python
cursor_mysql.execute("""
CREATE TABLE IF NOT EXISTS bookings (
  booking_id CHAR(50) PRIMARY KEY,
  booking_date DATE,
  booking_time TIME,
  day_of_week CHAR(50) ,
  is_weekend BOOLEAN DEFAULT 0,
  hour_of_day INT(10),
  city CHAR(50) ,
  pickup_location CHAR(50),
  drop_location  CHAR(50),
  vehicle_type CHAR(50),
  ride_distance_km DECIMAL(18,3),
  estimated_ride_time_min DECIMAL(18,3),
  actual_ride_time_min DECIMAL(18,3),
  traffic_level CHAR(50),
  weather_condition CHAR(50),
  base_fare DECIMAL(18,3),
  surge_multiplier DECIMAL(18,3),
  booking_value DECIMAL(18,3),
  booking_status CHAR(50),
  incomplete_ride_reason CHAR(50),
  customer_id CHAR(50)  NOT NULL,
  driver_id CHAR(50) NOT NULL);
""")
conn_mysql.commit()
print("Table 'bookings' created successfully in MySQL!")
```

```python
print("bookings columns:", len(bookings.columns))
print(bookings.columns)
#print("Placeholders in query:", insert_query.count('%s'))
```

```python
print (bookings.dtypes)
```

```python
# Build query AFTER cleaning columns
cols = ", ".join([f"`{col}`" for col in bookings.columns])
placeholders = ", ".join(["%s"] * len(bookings.columns))
insert_query = f"INSERT INTO `bookings` ({cols}) VALUES ({placeholders})"

print("Final query:", insert_query) # VERIFY no `nan` in here

chunk_size = 2000
for start in range(0, len(bookings), chunk_size):
    chunk = bookings.iloc[start:start+chunk_size]
    data = chunk.where(pd.notnull(chunk), None).values.tolist()
    cursor_mysql.executemany(insert_query, data)
    conn_mysql.commit()
    print(f"Inserted {start+len(chunk)}/{len(bookings)} rows")
```
